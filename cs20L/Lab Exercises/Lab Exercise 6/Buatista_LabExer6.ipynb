```python
#Import Libs
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score
from sklearn.preprocessing import StandardScaler
```

```python
#Load the data

df = pd.read_csv('covid_toy.csv')
df
```

```python
#Getting to know the dataset
df.info()
df.describe()
```

```python

X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values
```

```python
#Imputing
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
imputer.fit(X[:, 2:3])
X[:, 2:3] = imputer.transform(X[:, 2:3])
```

```python
print(X)
```

```python
#convert categorical data into numerical data using onehat
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [4])], remainder='passthrough')
X = np.array(ct.fit_transform(X))
```

```python

from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder()
X[:, 7:8] = oe.fit_transform(X[:, 7:8])
print(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20, random_state = 0)
X_train
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = le.fit_transform(y)
print(y)
```

```python
# prompt: #Feature scaling
from sklearn.preprocessing import StandardScaler

# Identify columns with non-numerical data
non_numeric_cols = [i for i in range(X_train.shape[1]) if isinstance(X_train[0, i], str)]

# Convert non-numerical columns to numerical (assuming they are categorical)
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
for col in non_numeric_cols:
    X_train[:, col] = le.fit_transform(X_train[:, col])
    X_test[:, col] = le.transform(X_test[:, col])  # Use the same encoder for consistency

sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
print(X_train)
print(X_test)
```

```python
from sklearn.tree import DecisionTreeClassifier # DecisionTreeClassifier

classifier = DecisionTreeClassifier(criterion = 'entropy', random_state = 0)
classifier.fit(X_train, y_train)
```

```python
y_pred = classifier.predict(X_test)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

```python
#getting the confusion matrix of the random forest model with its classification report
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
cm = confusion_matrix(y_test, y_pred)
confusion = pd.DataFrame(cm, index = ['Has Covid', 'No Covid'], columns = ['Predicted Has Covid ', 'Predicted No Covid'])
```

```python
print(confusion)
```

```python
#Predicting the accuracy
print(accuracy_score(y_test, y_pred))
#F1 scores
print(classification_report(y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier #Forest classifier
classifier = RandomForestClassifier(n_estimators = 10, criterion = 'entropy', random_state = 0)
classifier.fit(X_train, y_train)
```

```python
y_pred = classifier.predict(X_test)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
cm = confusion_matrix(y_test, y_pred)
confusion = pd.DataFrame(cm, index = ['Has Covid', 'No Covid'], columns = ['Predicted Has Covid ', 'Predicted No Covid'])
```

```python
print(confusion)
```

```python
#Predicting the accuracy
print(accuracy_score(y_test, y_pred))
#F1 Scores
print(classification_report(y_test, y_pred))
```

```python

# For Decision Tree
print("Decision Tree Performance:")
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))

# For Random Forest
print("\nRandom Forest Performance:")
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))

```

Both the Decision Tree and Random Forest models were trained and evaluated on the provided dataset. The Random Forest model slightly outperformed the Decision Tree in terms of overall accuracy. Precision, recall, and F1-score were used to evaluate the performance for each class (Has Covid and No Covid). Both models demonstrated reasonably good performance, with the Random Forest exhibiting marginally better results across most metrics.
