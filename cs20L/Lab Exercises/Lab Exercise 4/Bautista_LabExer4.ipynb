```python
#Importing the libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
#Importing the dataset
dataset = pd.read_csv('tvmarketing.csv')
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, 1].values

print(dataset)
```

```python
#Splitting the data dataset into the training set and test set
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20, random_state = 0)
```

```python
#Training the simple linear regression model on tyhe training set
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

```python
#Predict
y_pred = regressor.predict(X_test)
```

```python
#visualize the data
plt.scatter(X_train, y_train, color = 'red')
plt.plot(X_train, regressor.predict(X_train), color = 'blue')
plt.title('TV vs Sales (Training set)')
plt.xlabel('TV')
plt.ylabel('Sales')
plt.show()
```

```python
#Visualising the test set result
plt.scatter(X_test, y_test, color = 'red')
plt.plot(X_train, regressor.predict(X_train), color = 'blue')
plt.title('TV vs Sales (Training set)')
plt.xlabel('TV')
plt.ylabel('Sales')
plt.show()
```

```python
predicted_sales = regressor.predict([[255]])
print(f"Predicted sales for TV advertising budget of 255: {predicted_sales[0]:.2f}")
```


# Based on the scatter plot, we can observe a positive correlation between TV marketing and sales. This means that as TV marketing expenditure increases, sales also tend to increase. This is further confirmed by the linear regression line, which shows an upward trend.
# The strength of the relationship can be quantified using the correlation coefficient, which is a measure of how closely the data points are clustered around the regression line. In this case, the correlation coefficient is likely to be high, indicating a strong positive relationship.
# This suggests that TV marketing is an effective way to boost sales. However, it is important to note that other factors, such as product quality and price, may also influence sales.
# To determine the accuracy of the prediction, we can compare the predicted sales for a given TV advertising budget with the actual sales data. If the predictions are close to the actual data, then the model can be considered accurate.
# Overall, the analysis suggests that TV marketing has a positive impact on sales, and the model can be used to predict sales based on TV advertising expenditure.
