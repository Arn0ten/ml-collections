# Stemming and Cleaning the Data

Libraries of NLTK


```python
import re,string,nltk
import tensorflow as tf
from nltk.corpus import stopwords
from nltk.corpus import wordnet
from nltk.stem import PorterStemmer,WordNetLemmatizer
from spellchecker import SpellChecker
import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer
```

Stemming

---



```python
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))
```

Cleaning the data to reduce and improve the performance

---



```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r'https?://\S+|www\.\S+', 'URL', text)
    text = re.sub(r'<.*?>', '', text)
    text = re.sub(r'@\w+', 'USER', text)
    text = re.sub(r'\d+', '', text)
    text = re.sub(rf"[{re.escape(string.punctuation)}]", '', text)
    text = ' '.join([lemmatizer.lemmatize(word) for word in text.split() if word not in stop_words])
    return text
```

Applying the cleaning proccess


---



```python
data['cleaned_messages'] = data['messages'].apply(clean_text)
```

# Tokenization & Padding

Tokenization

---



```python
tokenizer = Tokenizer(num_words=5000, oov_token='<OOV>')

tokenizer.fit_on_texts(data['cleaned_messages'])

sequences = tokenizer.texts_to_sequences(data['cleaned_messages'])
```

Padding

---



```python
padded_sequences = pad_sequences(sequences, maxlen=100, padding='post')
```

```python
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt

def wordcount_gen(df, category, sentiment_column='labels'): # Change sentiment_column to 'labels'

    # Filter based on 'labels' column
    filtered_data = df[df[sentiment_column] == 1] if category == 'Positive' else df[df[sentiment_column] == -1] if category == 'Negative' else df[df[sentiment_column] == 0]
    # Combine messages for wordcloud
    combined_tweets = " ".join(filtered_data['messages'])

    wc = WordCloud(background_color='black',
                    max_words=50,
                    stopwords=STOPWORDS)
    plt.figure(figsize=(10, 10))
    plt.imshow(wc.generate(combined_tweets))
    plt.title('{} Sentiment Words'.format(category), fontsize=20)
    plt.axis('off')
    plt.show()
```

BAR PLOT OF WORD FREQUENCIES

---



```python
def plot_word_frequencies(text, title):
    word_freq = nltk.FreqDist(text.split())
    common_words = word_freq.most_common(20)
    words, freqs = zip(*common_words)
    plt.figure(figsize=(10, 6))
    sns.barplot(x=list(words), y=list(freqs), palette='viridis')
    plt.title(title)
    plt.xlabel('Words')
    plt.ylabel('Frequency')
    plt.xticks(rotation=45)
    plt.show()
# Step 3: Plot bar plots of word frequencies
plot_word_frequencies(' '.join(data['cleaned_messages']), 'Word Frequencies After Cleaning')
```

# Build the TensorFlow model

Recurrent Neural Network (RNN), specifically a Bidirectional LSTM with a Deep learning architecture.

128-dimensional vecto- o
input sequence

```python
model = Sequential([
    Embedding(input_dim=5000, output_dim=128, input_length=100),
    Bidirectional(LSTM(64, return_sequences=True)),
    Dropout(0.5),
    Bidirectional(LSTM(32)),
    Dense(64, activation='relu'),
    Dropout(0.5),
    Dense(3, activation='softmax')
])
```

Compile the model

---



```python
model.compile(loss='sparse_categorical_crossentropy', optimizer=Adam(learning_rate=0.001), metrics=['accuracy'])
```

Train the model

---



```python
history = model.fit(X_train, y_train, validation_data=(X_test, y_test), epochs=10, batch_size=64)
```

# Model Evaluation

```python
y_pred = np.argmax(model.predict(X_test), axis=1)
```

f1 score

---



```python
print(classification_report(y_test, y_pred, target_names=['Negative', 'Neutral', 'Positive']))
```

Confusion matrix

# Example Catching user input and analyzing.

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer

sid = SentimentIntensityAnalyzer()

def analyze_sentiment(text):
  cleaned_text = clean_text(text)
  sequence = tokenizer.texts_to_sequences([cleaned_text])
  padded_sequence = pad_sequences(sequence, maxlen=100, padding='post')
  prediction = np.argmax(model.predict(padded_sequence), axis=1)[0]
  sentiment_mapping = {0: 'Negative', 1: 'Neutral', 2: 'Positive'}
  predicted_sentiment = sentiment_mapping[prediction]
  return predicted_sentiment

def verify_with_vader(text):
  vader_score = sid.polarity_scores(text)['compound']
  if vader_score >= 0.05:
    return "Positive"
  elif vader_score <= -0.05:
    return "Negative"
  else:
    return "Neutral"
```

Get user input

---



```python
user_input = input("Enter text to analyze: ")

# Analyze sentiment using your model and VADER
model_sentiment = analyze_sentiment(user_input)
vader_sentiment = verify_with_vader(user_input)

# Print results
print("Model Sentiment:", model_sentiment)
print()
print("VADER Sentiment:", vader_sentiment)
```
