```python
#Example 1 Python SVM Classification
from sklearn import svm
x = [ [170, 70, 10], [180, 80, 12], [170, 65, 8], [160,55,7], [172, 72, 9], [182, 82, 11] ]

y = [0,0,1,1, 0, 0] #Gender 0: Male, 1: female, added corresponding labels

clf = svm.SVC()
clf.fit(x,y)

p = clf.predict([[169,60,7]])
print(p)
```

```python
#Example 2 Python SVM Iris Classification
from sklearn import svm, datasets

iris=datasets.load_iris()


x = iris.data[:, :2]
y = iris.target

clf = svm.SVC()
clf.fit(x, y)

p = clf.predict([[5.4, 3.2]])
print(p)
```

```python
#Example 3 Python SVM Iris CSV CLassification
from sklearn import svm, datasets
import pandas as pd

df = pd.read_csv('IRIS2.csv')
x=df.values[:,:2]
s=df['species']
d = dict([(y, x) for x, y in enumerate(sorted(set(s)))])
y = [d[x] for x in s]

clf = svm.SVC()
clf.fit(x, y)

print(d)
p = clf.predict([[5.4, 3.2]])
print(p)
```
